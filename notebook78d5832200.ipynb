```python
# ==============================================================================
#  ENTERPRISE OPERATIONAL INTELLIGENCE PLATFORM - BULLETPROOF AUTOMATED PIPELINE
# ==============================================================================

# Step 1: HARDWARE DEPLOYMENT MANAGEMENT & CACHE FLUSHING
import os
import gc
import torch

print("Executing volatile memory sweeps and performance tuning...")
os.environ["PYTORCH_ALLOC_CONF"] = "expandable_segments:True"

# Prevents Hugging Face Trainer from spinning up unstable Multi-GPU DataParallel processes
torch.cuda.device_count = lambda: 1

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    print(f"Hardware Verified: Single-Core target emulation locked onto device: {torch.cuda.get_device_name(0)}")
else:
    raise RuntimeError("CRITICAL ERROR: No GPU accelerator detected. Ensure Kaggle Settings panel has GPU enabled before running.")

# Step 2: Synchronize Framework Dependencies Cleanly
import subprocess
print("Synchronizing corporate package libraries...")
subprocess.run(['pip', 'install', 'transformers', 'datasets', 'accelerate', 'peft', 'bitsandbytes', 'huggingface_hub', '-q'])

# Step 3: Secure Native Authentication via User Secrets Tracker
import json
import random
from transformers import AutoModelForCausalLM, AutoTokenizer, Trainer, TrainingArguments, DataCollatorForSeq2Seq, BitsAndBytesConfig
from peft import LoraConfig, get_peft_model, TaskType, PeftModel, prepare_model_for_kbit_training
from datasets import Dataset
from huggingface_hub import HfApi, login

print("Loading enterprise security tokens from environment secrets...")
try:
    from kaggle_secrets import UserSecretsClient
    user_secrets = UserSecretsClient()
    hf_token = user_secrets.get_secret("HF_TOKEN")
    if hf_token:
        login(token=hf_token)
        os.environ["HF_TOKEN"] = hf_token
        print("Success: Hugging Face Write Token successfully validated and locked.")
    else:
        raise ValueError("CRITICAL: 'HF_TOKEN' is missing or toggled off inside Kaggle Add-ons -> Secrets.")
except Exception as e:
    raise RuntimeError(f"Authentication Blocked: {e}")

# ------------------------------------------------------------------------------
# DEPLOYMENT NAMESPACE INTERFACE - REALIGNED FOR USER PERMISSIONS
# ------------------------------------------------------------------------------
HF_USERNAME = "my-ai-stack"  # 🚀 LOCKED TO ORGANIZATIONAL NAMESPACE TO PREVENT 403
BASE_REPO_NAME = "coffee-shop-ops-executive"
model_name = "Qwen/Qwen2.5-0.5B"

ADAPTER_REPO_ID = f"{HF_USERNAME}/{BASE_REPO_NAME}-adapters"
MERGED_REPO_ID  = f"{HF_USERNAME}/{BASE_REPO_NAME}-merged"
QUANT_REPO_ID   = f"{HF_USERNAME}/{BASE_REPO_NAME}-quantized-4bit"
# ------------------------------------------------------------------------------

# Step 4: Parse Base Data Tensors
dataset_path = '/kaggle/input/datasets/walidsobhie/training-data-full-jsonl/training_data_full.jsonl'
if not os.path.exists(dataset_path):
    found_files = []
    for root, dirs, files in os.walk('/kaggle/input'):
        for file in files:
            if file.endswith('.jsonl'):
                found_files.append(os.path.join(root, file))
    if found_files:
        dataset_path = found_files[0]
        print(f"Dataset active at directory target: {dataset_path}")
    else:
        raise FileNotFoundError("Verify your training dataset asset is correctly attached to your Kaggle instance.")

base_data = []
with open(dataset_path, 'r') as f:
    for line in f:
        line = line.strip()
        if line and line not in ['[', ']', ',']:
            try:
                loaded = json.loads(line)
                if isinstance(loaded, dict):
                    base_data.append(loaded)
            except:
                pass

print(f"Base Ingestion Complete: {len(base_data)} raw records loaded safely.")

# ==============================================================================
#  DATA AMPLIFICATION LAYER (6x Combinatorial Matrix)
# ==============================================================================
print("Executing combinatorial data amplification engine...")
expanded_corporate_matrix = []

for item in base_data:
    inp = item.get('input', {})
    out = item.get('output', {})
    sop = str(inp.get('sop_reference', 'GLOBAL-SOP')).upper()
    
    if "FIN" in sop:
        pillar = "Financial Controls & Governance"
        section = "SOP-FIN-042: Yield & Material Spoilage Regulation"
        threshold = "Waste must remain < 2.0% of total regional procurement volume."
        ebitda_risk = "High"
        cogs_bps = random.randint(120, 260)
        payroll_var = 0.0
    elif "LAB" in sop:
        pillar = "Labor Allocation & Workforce Efficiency"
        section = "SOP-LAB-015: Shift Overlap & Optimization Planning"
        threshold = "Total payroll spend must not exceed 24.0% of localized trailing revenue metrics."
        ebitda_risk = "Critical"
        cogs_bps = 0
        payroll_var = round(random.uniform(8.0, 18.5), 1)
    elif "OPS" in sop or "STRAT" in sop:
        pillar = "Brand Standards & Customer Experience Excellence"
        section = "SOP-OPS-100: Experiential Quality & Engagement Mandate"
        threshold = "Net Promoter Score (NPS) must maintain a baseline tier of > 78%"
        ebitda_risk = "Medium"
        cogs_bps = random.randint(20, 60)
        payroll_var = round(random.uniform(2.0, 6.0), 1)
    else:
        pillar = "Supply Chain Integrity & Facilities Governance"
        section = "SOP-SUP-202: Perishable Asset Ingestion Profiles"
        threshold = "Product shelf-life availability upon delivery must not drop below 85% of total lifecycle."
        ebitda_risk = "High"
        cogs_bps = random.randint(50, 110)
        payroll_var = 0.0

    archetypes = ["SYSTEMIC CLUSTER GAP", "LOCAL OUTLIER BREACH", "STRATEGIC RECOVERY INITIATIVE"]
    context_prefixes = ["CRITICAL ESCALATION", "QUARTERLY PERFORMANCE AUDIT", "OPERATIONAL VARIANCE DETECTED"]
    
    for i in range(6):
        arch_type = archetypes[i % 3]
        prefix = context_prefixes[i % 3]
        
        adjusted_cogs = max(0, cogs_bps + random.randint(-15, 15)) if cogs_bps > 0 else 0
        adjusted_payroll = max(0.0, payroll_var + round(random.uniform(-1.5, 1.5), 1)) if payroll_var > 0.0 else 0.0
        
        mutated_input = {
            "archetype": arch_type,
            "role": inp.get('role', 'Corporate Operation Manager'),
            "context": f"{prefix}: {inp.get('context', '')}",
            "stores_involved": inp.get('stores_involved', ['Global Cluster']),
            "sop_reference": sop,
            "sop_taxonomy": { "pillar": pillar, "section": section, "tolerance_threshold": threshold },
            "financial_vectors": { "cogs_impact_basis_points": adjusted_cogs, "ebitda_risk": ebitda_risk, "payroll_variance_percentage": adjusted_payroll },
            "scenario": inp.get('scenario', 'Standard baseline operating state.'),
            "goal": inp.get('goal', 'Align local indicators back to target tolerances.')
        }
        
        expanded_corporate_matrix.append({ "input": mutated_input, "output": out })

print(f"Matrix Expanded: {len(expanded_corporate_matrix)} unique training files synthesized.")

# ==============================================================================
#  CHATML SYNTAX COMPILER
# ==============================================================================
def to_corporate_chatml(item):
    inp = item.get('input', {})
    out = item.get('output', {})
    tax = inp.get('sop_taxonomy', {})
    fin = item.get('financial_vectors', {}) if item.get('financial_vectors') else inp.get('financial_vectors', {})
    stores_str = ', '.join(inp.get('stores_involved', []))
    
    system_prompt = "You are the Corporate Operations AI Engine. Your function is to enforce brand standards and protect corporate EBITDA."
    
    user_query = f"""[ENTERPRISE OPERATIONAL AUDIT REQUEST]
- TAXONOMY PILLAR: {tax.get('pillar')}
- COMPLIANCE CODE: {inp.get('sop_reference')} ({tax.get('section')})
- SYSTEMIC THRESHOLD: {tax.get('tolerance_threshold')}
- ARCHETYPE: {inp.get('archetype')}
- ENTITIES INVOLVED: {stores_str}
- FISCAL ANOMALY PROFILE: EBITDA Risk: {fin.get('ebitda_risk')} | COGS Shift: +{fin.get('cogs_impact_basis_points', 0)} bps | Payroll Variance: {fin.get('payroll_variance_percentage', 0.0)}%
- ESCALATION CASE BRIEF: {inp.get('context')} - {inp.get('scenario')}
- TARGET STRATEGIC METRIC: {inp.get('goal')}"""

    assistant_response = f"""### STRATEGIC AUDIT & MARGIN DIAGNOSTIC
{out.get('analysis')}

### ENTERPRISE INTERVENTION DIRECTIVE
{out.get('action_plan')}

### CORPORATE FISCAL & PERFORMANCE PROJECTIONS
{out.get('kpi_impact')}"""

    text = f"<|im_start|>system\n{system_prompt}<|im_end|>\n<|im_start|>user\n{user_query}<|im_end|>\n<|im_start|>assistant\n{assistant_response}<|im_end|>"
    return {"text": text}

train_data = [to_corporate_chatml(item) for item in expanded_corporate_matrix]

# ==============================================================================
#  QUANTIZED BASE MODEL CONFIGURATION
# ==============================================================================
print(f"Loading Base Structural Tokenizer: {model_name}...")
tokenizer = AutoTokenizer.from_pretrained(model_name)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16
)

print("Materializing compressed 4-bit operational base weights with Eager Attention...")
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=quantization_config,
    device_map={"":"cuda:0"}, 
    attn_implementation="eager"
)

model = prepare_model_for_kbit_training(model)

peft_config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    r=16, 
    lora_alpha=32,
    lora_dropout=0.05,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"]
)

model = get_peft_model(model, peft_config)
model.config.use_cache = False 
model.gradient_checkpointing_enable()

# ==============================================================================
#  TOKENIZE & DATASET PARTITIONING
# ==============================================================================
def tokenize_func(sample):
    tokenized = tokenizer(sample['text'], max_length=2048, truncation=True)
    tokenized["labels"] = tokenized["input_ids"].copy()
    return tokenized

dataset = Dataset.from_list(train_data).map(tokenize_func, batched=False).train_test_split(test_size=0.05)
train_ds = dataset['train']
eval_ds = dataset['test']

print(f"Data partitioning synchronized. Training: {len(train_ds)} matrices | Eval: {len(eval_ds)} matrices.")

data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model, padding=True)

# ==============================================================================
#  STABLE REAL-TIME NETWORK BACKUP CONTROLS
# ==============================================================================
training_args = TrainingArguments(
    output_dir="/kaggle/working/local-checkpoints",
    per_device_train_batch_size=1,          
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=64,         
    gradient_checkpointing=True,            
    dataloader_num_workers=0,               
    num_train_epochs=3, 
    learning_rate=2e-4,                     
    weight_decay=0.01, 
    warmup_steps=150,                       
    logging_steps=10,                       
    report_to="none",
    
    # 🚀 REAL-TIME HUB STREAMING CONFIGURATION
    save_steps=100,                         
    eval_strategy="steps",
    eval_steps=100,
    save_total_limit=2,                     
    push_to_hub=True,                       
    hub_model_id=ADAPTER_REPO_ID,           
    hub_strategy="checkpoint",              
    hub_token=os.environ["HF_TOKEN"]        
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_ds,
    eval_dataset=eval_ds,
    data_collator=data_collator,
)

# Pre-allocates the repository container on Hugging Face before step 1 initiates
trainer.create_model_card()

print("Triggering fine-tuning sequence window loop...")
trainer.train()

# Final local staging snapshot
local_adapter_path = "/kaggle/working/saved-adapters"
trainer.model.save_pretrained(local_adapter_path)
tokenizer.save_pretrained(local_adapter_path)

# ==============================================================================
#  AUTOMATED MULTI-REPO EXPORT PIPELINE
# ==============================================================================
print("\n" + "="*60)
print("ENGAGING HUGGING FACE FINAL DISTRIBUTION & PACKAGING SEQUENCE")
print("="*60)

# REPO 1: Final Adapter Layer Push
print(f"Uploading Variant 1/3 [Lightweight Adapter Model Layer] to: {ADAPTER_REPO_ID}...")
trainer.model.push_to_hub(ADAPTER_REPO_ID, private=False)
tokenizer.push_to_hub(ADAPTER_REPO_ID, private=False)

del model
del trainer
gc.collect()
torch.cuda.empty_cache()

# REPO 2: Fused Unquantized Base Model Parameter Merge
print("\nReloading clean base parameters to execute matrix merge math scales...")
base_model_clean = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="cpu" 
)

print(f"Fusing adapter matrices from {local_adapter_path} permanently into baseline layers...")
merged_model = PeftModel.from_pretrained(base_model_clean, local_adapter_path)
merged_model = merged_model.merge_and_unload()

print(f"Uploading Variant 2/3 [Standalone Fused Precision Model] to: {MERGED_REPO_ID}...")
merged_model.push_to_hub(MERGED_REPO_ID, private=False)
tokenizer.push_to_hub(MERGED_REPO_ID, private=False)

# REPO 3: Production 4-bit Optimized Compressed Edge Package
print("\nClearing parameters to isolate final Quantized compression sequence...")
del base_model_clean
del merged_model
gc.collect()
torch.cuda.empty_cache()

print("Reloading merged repository weights under specialized 4-bit configuration matrix parameters...")
quantized_deploy_model = AutoModelForCausalLM.from_pretrained(
    MERGED_REPO_ID,
    quantization_config=quantization_config,
    device_map={"":"cuda:0"}
)

print(f"Uploading Variant 3/3 [Production-Grade Edge Quantized 4-bit Model] to: {QUANT_REPO_ID}...")
quantized_deploy_model.push_to_hub(QUANT_REPO_ID, private=False)
tokenizer.push_to_hub(QUANT_REPO_ID, private=False)

print("\n" + "="*60)
print("SUCCESS: ALL 3 ENTERPRISE OPERATIONS MODEL ARTIFACTS REPOS LIVE ON THE HUB")
print(f"1. LoRA Layer Set Checkpoint: {ADAPTER_REPO_ID}")
print(f"2. Unified Fused Core Asset:  {MERGED_REPO_ID}")
print(f"3. 4-bit Production Edge QT:  {QUANT_REPO_ID}")
print("=" * 60)
```
